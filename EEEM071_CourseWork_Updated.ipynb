## 4.1 Collect All Results

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import re
from glob import glob

# Set style for all plots
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette("husl")

log_base = "/scratch/HS400/CV/logs/"

def parse_log(log_dir):
    """Parse training log to extract metrics."""
    log_file = os.path.join(log_dir, "log_train.txt")
    result = {'mAP': None, 'Rank-1': None, 'Rank-5': None, 'Rank-10': None, 'Rank-20': None}
    
    if os.path.exists(log_file):
        with open(log_file, 'r') as f:
            content = f.read()
            # Parse metrics
            map_match = re.search(r'mAP:\s*([\d.]+)%?', content)
            r1_match = re.search(r'Rank-1\s*[:\s]+([\d.]+)%?', content)
            r5_match = re.search(r'Rank-5\s*[:\s]+([\d.]+)%?', content)
            r10_match = re.search(r'Rank-10\s*[:\s]+([\d.]+)%?', content)
            r20_match = re.search(r'Rank-20\s*[:\s]+([\d.]+)%?', content)
            
            if map_match: result['mAP'] = float(map_match.group(1))
            if r1_match: result['Rank-1'] = float(r1_match.group(1))
            if r5_match: result['Rank-5'] = float(r5_match.group(1))
            if r10_match: result['Rank-10'] = float(r10_match.group(1))
            if r20_match: result['Rank-20'] = float(r20_match.group(1))
    
    return result

# Collect all results
results = []
for exp_dir in sorted(glob(os.path.join(log_base, "sec*"))):
    exp_name = os.path.basename(exp_dir)
    metrics = parse_log(exp_dir)
    metrics['Experiment'] = exp_name
    results.append(metrics)

df = pd.DataFrame(results)
df = df[['Experiment', 'mAP', 'Rank-1', 'Rank-5', 'Rank-10', 'Rank-20']]
print("All Results Collected:")
print(df.to_string(index=False))
print(f"\nTotal experiments: {len(df)}")
```

### Graph 11: Training Loss Curves

```python
def parse_training_loss(log_dir):
    """Extract epoch-wise training loss from log file."""
    log_file = os.path.join(log_dir, "log_train.txt")
    epochs, losses = [], []
    
    if os.path.exists(log_file):
        with open(log_file, 'r') as f:
            for line in f:
                # Match patterns like "Epoch [1/60]" and "Loss: 5.1234"
                epoch_match = re.search(r'Epoch\s*\[?(\d+)', line)
                loss_match = re.search(r'Loss[:\s]+([\d.]+)', line)
                if epoch_match and loss_match:
                    epochs.append(int(epoch_match.group(1)))
                    losses.append(float(loss_match.group(1)))
    
    return epochs, losses

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Architecture loss curves
ax1 = axes[0]
arch_experiments = ['sec1_mobilenet', 'sec1_resnet50', 'sec1_resnet18', 'sec1_vgg16']
colors = plt.cm.tab10(np.linspace(0, 1, len(arch_experiments)))

for idx, exp in enumerate(arch_experiments):
    exp_dir = os.path.join(log_base, exp)
    epochs, losses = parse_training_loss(exp_dir)
    if epochs and losses:
        ax1.plot(epochs, losses, label=exp.replace('sec1_', '').upper(), 
                color=colors[idx], linewidth=2)

ax1.set_xlabel('Epoch', fontsize=12)
ax1.set_ylabel('Training Loss', fontsize=12)
ax1.set_title('Training Loss: Architecture Comparison', fontsize=12, fontweight='bold')
ax1.legend(loc='upper right')
ax1.grid(True, alpha=0.3)

# Augmentation loss curves
ax2 = axes[1]
aug_experiments = ['sec2_baseline', 'sec2_jitter', 'sec2_erase', 'sec2_combo']
colors = plt.cm.Set2(np.linspace(0, 1, len(aug_experiments)))

for idx, exp in enumerate(aug_experiments):
    exp_dir = os.path.join(log_base, exp)
    epochs, losses = parse_training_loss(exp_dir)
    if epochs and losses:
        ax2.plot(epochs, losses, label=exp.replace('sec2_', '').title(), 
                color=colors[idx], linewidth=2)

ax2.set_xlabel('Epoch', fontsize=12)
ax2.set_ylabel('Training Loss', fontsize=12)
ax2.set_title('Training Loss: Augmentation Comparison', fontsize=12, fontweight='bold')
ax2.legend(loc='upper right')
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig(f'{log_base}plot_training_curves.png', dpi=150)
plt.show()
print(f"Saved: {log_base}plot_training_curves.png")
```

### Graph 12: Convergence Plot (Loss vs mAP over Training)

```python
def parse_convergence_data(log_dir):
    """Extract both loss and mAP over epochs."""
    log_file = os.path.join(log_dir, "log_train.txt")
    data = {'epoch': [], 'loss': [], 'mAP': []}
    
    if os.path.exists(log_file):
        with open(log_file, 'r') as f:
            content = f.read()
            # Find evaluation blocks
            eval_blocks = re.findall(r'Epoch\s*\[?(\d+).*?Loss[:\s]+([\d.]+).*?mAP[:\s]+([\d.]+)', 
                                     content, re.DOTALL)
            for epoch, loss, mAP in eval_blocks:
                data['epoch'].append(int(epoch))
                data['loss'].append(float(loss))
                data['mAP'].append(float(mAP))
    
    return data

fig, ax1 = plt.subplots(figsize=(10, 6))

# Use best performing model for convergence analysis
best_exp = 'sec1_resnet50'
conv_data = parse_convergence_data(os.path.join(log_base, best_exp))

if conv_data['epoch']:
    color1 = 'steelblue'
    ax1.set_xlabel('Epoch', fontsize=12)
    ax1.set_ylabel('Training Loss', color=color1, fontsize=12)
    ax1.plot(conv_data['epoch'], conv_data['loss'], color=color1, linewidth=2, label='Loss')
    ax1.tick_params(axis='y', labelcolor=color1)
    
    ax2 = ax1.twinx()
    color2 = 'coral'
    ax2.set_ylabel('Validation mAP (%)', color=color2, fontsize=12)
    ax2.plot(conv_data['epoch'], conv_data['mAP'], color=color2, linewidth=2, label='mAP')
    ax2.tick_params(axis='y', labelcolor=color2)
    
    plt.title(f'Convergence Analysis: {best_exp}', fontsize=14, fontweight='bold')
    fig.legend(loc='upper center', bbox_to_anchor=(0.5, 0.95), ncol=2)
else:
    ax1.text(0.5, 0.5, 'Run experiments first to see convergence data', 
             ha='center', va='center', transform=ax1.transAxes)

plt.tight_layout()
plt.savefig(f'{log_base}plot_convergence.png', dpi=150)
plt.show()
```

### Graph 13: Radar Chart - Multi-Metric Architecture Comparison

```python
from math import pi

def create_radar_chart(df, title, filename):
    """Create radar chart for multi-metric comparison."""
    
    # Select metrics for radar
    metrics = ['mAP', 'Rank-1', 'Rank-5', 'Rank-10']
    available_metrics = [m for m in metrics if m in df.columns and df[m].notna().any()]
    
    if len(available_metrics) < 3 or len(df) == 0:
        print(f"Not enough data for radar chart: {title}")
        return
    
    # Number of variables
    num_vars = len(available_metrics)
    
    # Compute angle for each axis
    angles = [n / float(num_vars) * 2 * pi for n in range(num_vars)]
    angles += angles[:1]  # Complete the loop
    
    # Initialize figure
    fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True))
    
    # Colors
    colors = plt.cm.tab10(np.linspace(0, 1, len(df)))
    
    for idx, (_, row) in enumerate(df.iterrows()):
        values = [row[m] if pd.notna(row[m]) else 0 for m in available_metrics]
        values += values[:1]  # Complete the loop
        
        ax.plot(angles, values, 'o-', linewidth=2, label=row.get('Architecture', row.get('Experiment', f'Model {idx}')), color=colors[idx])
        ax.fill(angles, values, alpha=0.1, color=colors[idx])
    
    # Set labels
    ax.set_xticks(angles[:-1])
    ax.set_xticklabels(available_metrics, fontsize=11)
    ax.set_title(title, fontsize=14, fontweight='bold', y=1.08)
    ax.legend(loc='upper right', bbox_to_anchor=(1.3, 1.0))
    
    plt.tight_layout()
    plt.savefig(f'{log_base}{filename}', dpi=150, bbox_inches='tight')
    plt.show()

# Create radar chart for architectures
if len(arch_df) > 0:
    create_radar_chart(arch_df, 'Architecture Multi-Metric Comparison', 'plot_radar_arch.png')
else:
    print("Run architecture experiments first")
```

### Graph 16: Training Time & Efficiency Comparison

```python
def parse_training_time(log_dir):
    """Extract training time from log file."""
    log_file = os.path.join(log_dir, "log_train.txt")
    
    if os.path.exists(log_file):
        with open(log_file, 'r') as f:
            content = f.read()
            # Look for total time or time per epoch
            time_match = re.search(r'Total time[:\s]+([\d.]+)\s*(s|min|h)', content, re.IGNORECASE)
            epoch_time_match = re.search(r'Time[:\s]+([\d.]+)\s*s/epoch', content, re.IGNORECASE)
            
            if time_match:
                time_val = float(time_match.group(1))
                unit = time_match.group(2).lower()
                if unit == 'min': time_val *= 60
                elif unit == 'h': time_val *= 3600
                return time_val
            elif epoch_time_match:
                return float(epoch_time_match.group(1)) * 60  # Assume 60 epochs
    return None

# Collect training times
time_data = []
for _, row in arch_df.iterrows():
    exp_dir = os.path.join(log_base, row['Experiment'])
    train_time = parse_training_time(exp_dir)
    time_data.append({
        'Architecture': row['Architecture'],
        'Params (M)': row.get('Params (M)', 0),
        'mAP': row['mAP'],
        'Training Time (s)': train_time if train_time else np.nan
    })

time_df = pd.DataFrame(time_data)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot 1: Training time comparison
ax1 = axes[0]
if time_df['Training Time (s)'].notna().any():
    bars = ax1.bar(time_df['Architecture'], time_df['Training Time (s)'] / 60, 
                   color='steelblue', edgecolor='black')
    ax1.set_ylabel('Training Time (minutes)', fontsize=12)
else:
    # Estimate based on parameters
    estimated_times = time_df['Params (M)'] * 2  # Rough estimate
    bars = ax1.bar(time_df['Architecture'], estimated_times, color='lightsteelblue', edgecolor='black')
    ax1.set_ylabel('Estimated Training Time (minutes)', fontsize=12)
    ax1.text(0.5, 0.95, '* Estimated from model parameters', transform=ax1.transAxes, 
             ha='center', fontsize=9, style='italic')

ax1.set_xlabel('Architecture', fontsize=12)
ax1.set_title('Training Time Comparison', fontsize=12, fontweight='bold')
ax1.set_xticklabels(time_df['Architecture'], rotation=45, ha='right')
ax1.grid(axis='y', alpha=0.3)

# Plot 2: Efficiency (mAP per parameter)
ax2 = axes[1]
efficiency = time_df['mAP'] / time_df['Params (M)']
colors = ['green' if e == efficiency.max() else 'steelblue' for e in efficiency]
ax2.bar(time_df['Architecture'], efficiency, color=colors, edgecolor='black')
ax2.set_xlabel('Architecture', fontsize=12)
ax2.set_ylabel('Efficiency (mAP / Parameters)', fontsize=12)
ax2.set_title('Model Efficiency: Performance per Parameter', fontsize=12, fontweight='bold')
ax2.set_xticklabels(time_df['Architecture'], rotation=45, ha='right')
ax2.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.savefig(f'{log_base}plot_training_efficiency.png', dpi=150)
plt.show()
```

### Graph 19: Pareto Frontier - Efficiency vs Performance Trade-off

```python
def find_pareto_frontier(df, x_col, y_col, maximize_x=False, maximize_y=True):
    """Find Pareto optimal points."""
    pareto_points = []
    
    for i, row in df.iterrows():
        is_dominated = False
        for j, other in df.iterrows():
            if i == j:
                continue
            
            # Check if 'other' dominates 'row'
            if maximize_x and maximize_y:
                if other[x_col] >= row[x_col] and other[y_col] >= row[y_col]:
                    if other[x_col] > row[x_col] or other[y_col] > row[y_col]:
                        is_dominated = True
                        break
            elif not maximize_x and maximize_y:
                if other[x_col] <= row[x_col] and other[y_col] >= row[y_col]:
                    if other[x_col] < row[x_col] or other[y_col] > row[y_col]:
                        is_dominated = True
                        break
        
        if not is_dominated:
            pareto_points.append(i)
    
    return pareto_points

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Plot 1: Parameters vs mAP Pareto
ax1 = axes[0]
if len(arch_df) > 0 and 'Params (M)' in arch_df.columns:
    # Scatter all points
    ax1.scatter(arch_df['Params (M)'], arch_df['mAP'], s=150, c='steelblue', 
                edgecolors='black', linewidths=1.5, label='All Models', zorder=2)
    
    # Find and highlight Pareto frontier (minimize params, maximize mAP)
    pareto_idx = find_pareto_frontier(arch_df.reset_index(), 'Params (M)', 'mAP', 
                                       maximize_x=False, maximize_y=True)
    pareto_df = arch_df.iloc[pareto_idx].sort_values('Params (M)')
    
    if len(pareto_df) > 0:
        ax1.plot(pareto_df['Params (M)'], pareto_df['mAP'], 'r--', linewidth=2, 
                label='Pareto Frontier', zorder=1)
        ax1.scatter(pareto_df['Params (M)'], pareto_df['mAP'], s=200, c='red', 
                   marker='*', edgecolors='black', linewidths=1, label='Pareto Optimal', zorder=3)
    
    # Annotate points
    for _, row in arch_df.iterrows():
        ax1.annotate(row['Architecture'], (row['Params (M)'], row['mAP']),
                    xytext=(5, 5), textcoords='offset points', fontsize=9)
    
    ax1.set_xlabel('Parameters (Millions)', fontsize=12)
    ax1.set_ylabel('mAP (%)', fontsize=12)
    ax1.set_title('Pareto Frontier: Model Size vs Performance', fontsize=12, fontweight='bold')
    ax1.legend(loc='lower right')
    ax1.grid(True, alpha=0.3)

# Plot 2: Improvement over baseline
ax2 = axes[1]
if len(aug_df) > 0 and 'mAP Δ' in aug_df.columns:
    colors = ['green' if x > 0 else 'red' for x in aug_df['mAP Δ']]
    bars = ax2.barh(aug_df['Augmentation'], aug_df['mAP Δ'], color=colors, edgecolor='black')
    ax2.axvline(0, color='black', linewidth=1)
    ax2.set_xlabel('mAP Change from Baseline (%)', fontsize=12)
    ax2.set_title('Augmentation Impact: Improvement over Baseline', fontsize=12, fontweight='bold')
    ax2.grid(axis='x', alpha=0.3)
    
    # Add value labels
    for bar, val in zip(bars, aug_df['mAP Δ']):
        if pd.notna(val):
            ax2.text(val + 0.1 if val >= 0 else val - 0.5, bar.get_y() + bar.get_height()/2,
                    f'{val:+.1f}', va='center', fontsize=9)
elif len(aug_df) > 0:
    ax2.barh(aug_df['Augmentation'], aug_df['mAP'], color='steelblue', edgecolor='black')
    ax2.set_xlabel('mAP (%)', fontsize=12)
    ax2.set_title('Augmentation Performance', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.savefig(f'{log_base}plot_pareto_frontier.png', dpi=150)
plt.show()

print("\n" + "="*60)
print("ALL 19 VISUALIZATIONS GENERATED!")
print("="*60)
```
